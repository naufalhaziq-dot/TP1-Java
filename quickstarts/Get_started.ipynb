```
# Import library
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture

# Membuat dataset
X, y = make_blobs(
    n_samples=300,
    centers=3,
    random_state=42
)

# Model KMeans
kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

kmeans_labels = kmeans.fit_predict(X)

# Model Gaussian Mixture
gmm = GaussianMixture(
    n_components=3,
    random_state=42
)

gmm_labels = gmm.fit_predict(X)

# Visualisasi
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.scatter(
    X[:,0],
    X[:,1],
    c=kmeans_labels,
    cmap='viridis'
)
plt.title("KMeans Clustering")

plt.subplot(1,2,2)
plt.scatter(
    X[:,0],
    X[:,1],
    c=gmm_labels,
    cmap='viridis'
)
plt.title("Gaussian Mixture Model")

plt.tight_layout()
plt.show()
```
