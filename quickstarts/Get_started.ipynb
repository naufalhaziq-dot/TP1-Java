```
import statistics

class StatAnalyzer:
    def __init__(self, data):
        # Encapsulation (atribut private)
        self.__data = data

    def mean(self):
        return statistics.mean(self.__data)

    def median(self):
        return statistics.median(self.__data)

    def std(self):
        # Standar deviasi sampel
        return statistics.stdev(self.__data)

# Data yang diberikan
angka = [5, 7, 3, 9, 2]

# Membuat objek
analyzer = StatAnalyzer(angka)

# Menampilkan hasil
print("Data   :", angka)
print("Mean   :", analyzer.mean())
print("Median :", analyzer.median())
print("Std    :", round(analyzer.std(), 2))
```
